# Google Trends: AI Tools Comparison

Weekly worldwide Google Trends interest for ChatGPT, Gemini, and Microsoft Copilot, from late August 2023 to September 2024.

Questions:
1. Which tool has the most consistent growth in interest?
2. When does ChatGPT see its largest decline?
3. Which month has the highest average interest across all tools?

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the data (ai_tools_comparison.csv comes from the DataCamp lab)
trends = pd.read_csv('ai_tools_comparison.csv')
trends.head()
```

```python
# Parse dates, index by week, check for missing values
trends['week'] = pd.to_datetime(trends['week'])
trends = trends.set_index('week')
print(trends.isna().sum())

# Interest over time for all three tools
trends.plot(figsize=(12, 5), title='Google Trends interest over time')
plt.ylabel('Search interest')
plt.show()
```

```python
# Q1: consistency = lowest std of week over week percent change
pct = trends.pct_change().dropna()
print(pct.std())
most_consistent_tool = pct.std().idxmin()
print(most_consistent_tool)
```

```python
# Q2: largest single week drop in ChatGPT interest (ignoring the final partial week)
gpt_dip_week = trends['chatgpt'].diff().loc[:'2024-08-25'].idxmin()
gpt_dip = gpt_dip_week.strftime('%B %Y')
print(gpt_dip)
```

```python
# Q3: month with the highest average interest across all tools
best_month = trends.mean(axis=1).groupby(trends.index.month_name()).mean().idxmax()
print(best_month)
```

## Results

* Most consistent tool: **chatgpt** (week over week change std about 0.09, versus 0.36 for Gemini and 0.25 for Copilot)
* Largest ChatGPT decline: **December 2023**
* Month with highest average interest: **May**
